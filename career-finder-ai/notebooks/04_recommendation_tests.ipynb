# 04 – Recommendation Tests

End-to-end test of the recommendation engine using sample student profiles.

```python
import sys
sys.path.insert(0, '../backend')

from app.recommender import recommend_from_message
import pandas as pd
```

```python
sample_messages = [
    'I am a cybersecurity student in Dammam looking for remote COOP',
    'Data science student in Riyadh, prefer on-site internship',
    'AI student looking for any internship in Jeddah',
    'Computer engineering student, no city preference, hybrid mode',
]

for msg in sample_messages:
    print(f'\nMessage: {msg}')
    response = recommend_from_message(msg)
    print(f'Profile: {response.profile}')
    print('Top recommendations:')
    for i, opp in enumerate(response.recommendations, 1):
        print(f'  {i}. [{opp.score:.2f}] {opp.company} – {opp.title} ({opp.city}, {opp.work_mode})')
```

```python
# Build a summary table
rows = []
for msg in sample_messages:
    resp = recommend_from_message(msg)
    for rank, opp in enumerate(resp.recommendations, 1):
        rows.append({
            'message': msg[:40],
            'rank': rank,
            'company': opp.company,
            'title': opp.title,
            'score': opp.score,
        })

pd.DataFrame(rows)
```
